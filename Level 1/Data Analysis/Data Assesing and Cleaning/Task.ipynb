--- PHASE 1: STRUCTURAL CHECKS ---

S3. Duplicate Detection
    df.duplicated().sum()

--- PHASE 2: DISTRIBUTION ANALYSIS ---

D1. Descriptive Statistics
    df.describe()
    df['col'].mean()
    df['col'].median()

D2. Value Counts (for categoricals)
    df['col'].value_counts()
    df['col'].nunique()

D3. Min / Max Boundary Check
    df['col'].min()
    df['col'].max()
    df.nlargest(10, 'col')
    df.nsmallest(10, 'col')


--- PHASE 3: STATISTICAL OUTLIER DETECTION ---

N1. IQR Method (works on any distribution)
    Q1 = df['col'].quantile(0.25)
    Q3 = df['col'].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df['col'] < Q1 - 1.5*IQR) | (df['col'] > Q3 + 1.5*IQR)]

N2. Z-Score Method (for normal distributions)
    from scipy import stats
    z = (df['col'] - df['col'].mean()) / df['col'].std()
    outliers = df[z.abs() > 3]

N3. Percentile Thresholds
    df[df['col'] > df['col'].quantile(0.99)]
    df[df['col'] < df['col'].quantile(0.01)]


--- PHASE 4: LOGIC / SEMANTIC RULES ---

L1. Cross-Column Consistency Check
    df[(df['col1'] == 0) & (df['col2'] > 0)]
    df[(df['price'] == 0) & (df['ratings'] > 0)]

L2. Group-By Variance Check
    df.groupby('name')['price'].agg(['min', 'max', 'count'])
    # flag where min != max

L3. String Pattern / Format Check
    df['col'].str.contains(' ')
    df['col'].str.match(r'^[A-Z]')
    df['col'].str.len().value_counts()

L4. Date Boundary Check
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df[df['date'] > pd.Timestamp.today()]   # future dates
    df[df['date'].isnull()]                 # invalid dates
    df['date'].dt.year.value_counts()       # year distribution

## Handling Tidyness

```python
df['author'].split()
```

```python
import re

def fix_name(text):
    text = str(text)
    text = re.sub(r'([a-z])([A-Z])', r'\1 \2', text)       # camelCase split
    text = re.sub(r'([A-Z]+)([A-Z][a-z])', r'\1 \2', text) # acronym split
    return text.strip()

df['author_clean']   = df['author'].apply(fix_name)
df['narrator_clean'] = df['narrator'].apply(fix_name)
```

```python
df
```

```python
df.drop(columns=['author','narrator'],inplace=True)
```

```python
df.rename(columns={'author_clean':'author','narrator_clean':'narrator'},inplace=True)
```
