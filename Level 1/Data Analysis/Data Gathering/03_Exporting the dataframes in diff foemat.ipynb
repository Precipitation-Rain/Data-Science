# to_csv

```python
import pandas as pd
```

```python
df  = pd.read_csv(r"D:\Study\Data Science\Level 1\Frameworks\Pandas\Datasets\18_ipl_deliveries.csv")
```

```python
df
```

```python
new_df = df.groupby('batter')['batsman_run'].sum().sort_values(ascending=False).reset_index()
```

```python
new_df.to_csv('Batmsan runs in ipl.csv')
```

# to_excel

```python
new_df.to_excel('Batsman runs.xlsx')
```

```python
temp_df = df.pivot_table(index = 'batter' , columns = 'BowlingTeam' , values = 'batsman_run' , aggfunc = 'sum')
```

```python
temp_df.to_html('each batmsan runs against the each ipl team.html')
```

# to_json

```python
temp_df.to_json('each batsman runs in the ipl.json')
```

```python
df.groupby(['BattingTeam','batter'])['batsman_run'].sum().unstack().to_json('ipl.json')
```
