# Introductory

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
from sqlalchemy import create_engine
```

```python
conn = create_engine("mysql+pymysql://root:1771@localhost/ecommerce")
```

```python
category = pd.read_csv('Datasets\categories.csv')
customer = pd.read_csv('Datasets\customers.csv')
department = pd.read_csv('Datasets\departments.csv')
employee = pd.read_csv('Datasets\employees.csv')
order_item = pd.read_csv('Datasets\order_items.csv')
product = pd.read_csv('Datasets\products.csv')
order = pd.read_csv('Datasets\orders.csv')
```

# Upload one table 

```python
order.to_sql(
    name='orders',
    con=conn,
    if_exists='replace',
    index=False
)
```

# Upload multiple Tables/DF

```python
tables ={
    'category' : category,
    'customer' : customer,
    'department' : department,
    'employee' : employee,
    'order_item' : order_item,
    'product' : product,
    'order' : order
}
```

```python
for table_name , df in tables.items():

    df.to_sql(
        name = table_name,
        con = conn,
        if_exists = 'replace',
        index = False
    )
```
