# Getting all Data

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

page_url = "https://news.ycombinator.com/"  # renamed from url
headers = {"User-Agent": "Mozilla/5.0"}
new_df = pd.DataFrame()
page_count = 0

while True:
    response = requests.get(page_url, headers=headers)
    soup = BeautifulSoup(response.text, "html.parser")
    
    allcontent = soup.select("tr.athing.submission")
    l = []
    for item in allcontent:
        rank = item.find("span", class_='rank').text
        title = item.find("span", class_='titleline').find("a").text
        link = item.find("span", class_='titleline').find("a")["href"]  # actual article link
        site = item.find("span", class_='sitestr').text if item.find("span", class_='sitestr') else "N/A"  # None fix
        
        l.append({"Rank": rank, "Title": title, "Link": link, "Site": site})

    df1 = pd.DataFrame(l)

    allother = soup.find_all("span", class_='subline')
    m = []
    for item in allother:
        points = item.find("span").text
        author = item.find("a").text
        comments = item.find_all("a")[-1].text
        m.append({"Points": points, "Author": author, "Comments": comments})

    df2 = pd.DataFrame(m)

    df = pd.concat([df1, df2], axis=1)

    df['Rank'] = df['Rank'].str.replace('.', '')
    df['Points'] = df['Points'].str[:-6]
    df['Comments'] = df['Comments'].str[:-8].str.strip()

    new_df = pd.concat([new_df, df], ignore_index=True)

    next_button = soup.find("a", class_="morelink")
    if next_button:
        page_url = "https://news.ycombinator.com/" + next_button["href"]  # uses renamed variable
    else:
        break

    page_count += 1
    time.sleep(2)
```
