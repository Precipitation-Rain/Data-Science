````python
import plotly.graph_objects as go

# Example data (USA states)
states = ["CA", "TX", "NY", "FL"]
values = [100, 200, 150, 80]

fig = go.Figure()

# Choropleth (colored states)
fig.add_trace(go.Choropleth(
    locations=states,
    z=values,
    locationmode="USA-states",
    colorscale="Greens",
    showscale=False
))

# Add text labels
fig.add_trace(go.Scattergeo(
    locationmode="USA-states",
    locations=states,
    text=values,
    mode="text"
))

fig.update_layout(
    geo_scope='usa'
)

fig.show()
````

````python
import plotly.graph_objects as go

countries = ["IND", "USA", "BRA", "AUS", "CAN"]
values = [100, 200, 150, 80, 120]

fig = go.Figure()

# Choropleth (countries)
fig.add_trace(go.Choropleth(
    locations=countries,
    z=values,
    locationmode="ISO-3",
    colorscale="Greens",
    showscale=True
))

# Text labels on countries
fig.add_trace(go.Scattergeo(
    locations=countries,
    locationmode="ISO-3",
    text=values,
    mode="text"
))

fig.update_layout(
    geo=dict(
        scope="world",
        showland=True,
        landcolor="lightgray",
        showcountries=True
    )
)

fig.show()
````

# 🗺️ 1. Choropleth Map (Regions Colored)

👉 Used for countries / states

```python
import plotly.express as px
import pandas as pd

df = pd.DataFrame({
    "country": ["IND", "USA", "BRA"],
    "value": [100, 200, 150]
})

fig = px.choropleth(
    df,
    locations="country",
    locationmode="ISO-3",
    color="value",
    color_continuous_scale="Greens"
)

fig.show()
```

---

# 📍 2. Scatter Map (Points on Map)

👉 Plot locations using latitude & longitude

```python
import plotly.express as px

df = pd.DataFrame({
    "city": ["Mumbai", "Delhi"],
    "lat": [19.0760, 28.7041],
    "lon": [72.8777, 77.1025]
})

fig = px.scatter_mapbox(
    df,
    lat="lat",
    lon="lon",
    hover_name="city",
    zoom=4,
    mapbox_style="open-street-map"
)

fig.show()
```

---

# 🔵 3. Bubble Map (Size based on value)

👉 Like scatter but size shows importance

```python
fig = px.scatter_mapbox(
    df,
    lat="lat",
    lon="lon",
    size=[50, 100],
    hover_name="city",
    mapbox_style="open-street-map"
)

fig.show()
```

---

# 🌡️ 4. Density Heatmap (Heatmap on Map)

👉 Shows density (hot/cold regions)

```python
fig = px.density_mapbox(
    df,
    lat="lat",
    lon="lon",
    radius=20,
    mapbox_style="open-street-map"
)

fig.show()
```

---

# 🛣️ 5. Line Map (Routes / Paths)

👉 Used for travel routes

```python
import plotly.graph_objects as go

fig = go.Figure(go.Scattermapbox(
    mode="lines",
    lat=[19.0760, 28.7041],
    lon=[72.8777, 77.1025]
))

fig.update_layout(mapbox_style="open-street-map")
fig.show()
```

---

# 🌍 6. Scatter Geo (No Mapbox Required)

👉 Simple world map (like your image style)

```python
import plotly.graph_objects as go

fig = go.Figure(go.Scattergeo(
    locations=["IND", "USA"],
    locationmode="ISO-3",
    text=["India", "USA"],
    mode="markers+text"
))

fig.update_layout(geo=dict(scope="world"))
fig.show()
```

---

# 🗺️ 7. Choropleth + Text (Your Image Type)

👉 Colored regions + numbers on map

```python

fig = go.Figure()

fig.add_trace(go.Choropleth(
    locations=["USA", "IND"],

fig.add_trace(go.Scattergeo(
    locations=["USA", "IND"],
    locationmode="ISO-3",
    text=[200, 100],
    mode="text"
))

fig.show()
```

## 🎨 Common Map Styles

Plotly supports styles from **Mapbox** and **OpenStreetMap**:

* `"open-street-map"`
* `"carto-positron"`
* `"carto-darkmatter"`
* `"satellite"` (needs token)

---

# 🎯 Summary Table

| Map Type          | Use Case                |
| ----------------- | ----------------------- |
| Choropleth        | Country/State data      |
| Scatter Map       | Locations               |
| Bubble Map        | Size-based importance   |
| Density Map       | Heatmaps                |
| Line Map          | Routes                  |
| Scatter Geo       | Simple maps (no Mapbox) |
| Choropleth + Text | Dashboard-style maps    |

---

## 🚀 Final Tip

* Use **`scatter_mapbox`** → real-world interactive maps
* Use **`scatter_geo` / choropleth`** → simple/static maps
