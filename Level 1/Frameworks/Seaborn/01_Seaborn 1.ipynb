```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
# axes level plot => rectangle type image
sns.scatterplot(data = tips , x = "total_bill" , y = 'tip' , hue = 'sex' , style = 'time' , size = "size")
```

```python
# figure level plot => sqare type image
sns.relplot(data = tips , x = 'total_bill' , y = 'tip' , kind = 'scatter' , hue = 'sex' , style = 'time' , size = 'size')
```

## lineplot

```python
gap = px.data.gapminder()
```

```python
temp_df = gap[gap['country'] == 'India']
temp_df
```

```python
# axis level
sns.lineplot(data = temp_df , x = 'year' , y = 'lifeExp' )
```

```python
# fig level
sns.relplot(kind = 'line' , data = temp_df , x = 'year' , y = 'gdpPercap')
```

```python
# hue
temp = gap[gap['country'].isin(['India' , 'China', 'Pakistan'])]
temp
```

```python
# axes level plot
sns.lineplot(data = temp , x = 'year' , y = 'lifeExp' ,  hue = 'country' )
```

```python
# style
temp = gap[gap['country'].isin(['India' , 'France', 'Japan'])]
temp
```

```python
# figure level plot
sns.relplot(kind = 'line' ,data = temp , x = 'year' , y = 'gdpPercap' , hue = 'country' , size = 'continent')
```

```python
# facet plot -> figure level function -> work with relplot
# it will not work with scatterplot and lineplot
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter' , col = 'sex' , row = 'day' )
```

```python
# col_wrap
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter' , col = 'day' , col_wrap = 2)
```

## rugplot()

```python
# Rugplot

# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
```

```python
sns.displot(data = tips , x = 'total_bill' , kind = 'kde')
sns.rugplot(data = tips , x = 'total_bill' )
```

```python
# prefered way
sns.kdeplot(data = tips , x = 'total_bill')
sns.rugplot(data = tips , x = 'total_bill' )
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot 
# and then shows the count of observations within each rectangle with the fill color
```

```python
sns.histplot(tips , x = 'total_bill' , y = 'tip')
```

```python
sns.displot(tips , x = 'total_bill' , y = 'tip' , kind = 'hist')
```

```python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian
```

```python
sns.kdeplot(tips , x ='total_bill' , y = 'tip')
```

```python
sns.displot(tips , x ='total_bill' , y = 'tip' , kind = 'kde')
```

## matrix plot()

<h4>Heatmap</>

```python
# Plot rectangular data as a color-encoded matrix
```

```python
temp_df = gap.pivot(index ='country' , columns='year' , values='lifeExp')
```

```python
# heatmap => only axis level function , has no counter figire level function"
# you can apply all matplotlib properties on axes level function
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
# annot
```

```python
temp_df = gap[gap['continent'] =='Europe']
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(temp_df.pivot(index ='country' , columns='year' , values='lifeExp'), annot=True)
```

```python
# cmap
plt.figure(figsize=(15,10))
sns.heatmap(temp_df.pivot(index ='country' , columns='year' , values='lifeExp'), annot=True , cmap = 'winter')
```

```python
# linewidth
plt.figure(figsize=(15,10))
sns.heatmap(temp_df.pivot(index ='country' , columns='year' , values='lifeExp'), annot=True , cmap = 'winter' , linewidths=0.5)
```

```python
# Clustermap
# Plot a matrix dataset as a hierarchically-clustered heatmap.
# This function requires scipy to be available.
```

```python
iris = px.data.iris()
iris
```

```python
# plt.figure(figsize=(50,50))
sns.clustermap(iris.iloc[:,[0,1,2,3]] , cmap = 'viridis'  , linewidths = 0.5)
```

```python
sns.clustermap(temp_df.pivot(index ='country' , columns='year' , values='lifeExp'), annot=True , cmap = 'winter' , linewidths=0.5)
```
