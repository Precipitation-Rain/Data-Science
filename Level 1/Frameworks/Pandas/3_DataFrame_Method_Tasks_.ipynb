````python
teams = df["Team"].unique()
print(len(teams))   # total number of teams
````

````python
team_on_target_percentage = {}

for team in teams:
    team_df = df[df["Team"] == team]

    total_attempts = team_df["Total Attempts"].sum()
    total_on_target = team_df["On Target"].sum()

    if total_attempts > 0:  # safety check
        percentage = (total_on_target / total_attempts) * 100
    else:
        percentage = 0

    team_on_target_percentage[team] = round(percentage, 2)
````

````python
sorted_team_on_target_percentage = dict(
    sorted(
        team_on_target_percentage.items(),
        key=lambda item: item[1],
        reverse=True
    )
)
````
