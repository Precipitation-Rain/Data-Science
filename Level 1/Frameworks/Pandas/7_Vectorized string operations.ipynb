### date_range()

```python
# generate daily dates in given range
pd.date_range(start = '2026/2/18' , end = '2026/2/27')
```

```python
# alternative day
pd.date_range(start = '2026/2/18' , end = '2026/2/27',freq='2D')
```

```python
# buissness days
pd.date_range(start = '2026/2/18' , end = '2026/2/27',freq='B')
```

```python
# one week one day
pd.date_range(start = '2026/2/18' , end = '2026/2/27',freq='W-MON')
```

```python
# hourly data
pd.date_range(start = '2026/2/18' , end = '2026/2/27',freq='H')
```

```python
# 6h differnce
pd.date_range(start = '2026/2/18' , end = '2026/2/27',freq='6H')
```

```python
# month end
pd.date_range(start = '2026/2/18' , end = '2026/3/27',freq='ME')
```

```python
# year end
pd.date_range(start = '2026/2/18' , end = '2029/2/27',freq='YE')
```

```python
# using periods => alternative of end parameter
pd.date_range(start = '2026/2/18' , periods=27,freq='2D')
```

### to_datetime() 
converts an existing objects to pandas timestamp/datetimeindex object

```python
s = pd.Series(['2026/1/18','2026/1/19','2026/1/12','2026/1/13','2026/1/14'])
```

```python
pd.to_datetime(s)
```

```python
pd.to_datetime(s).dt.year
```

```python
# with errors
s = pd.Series(['2026/1/18','2026/1/19','2026/1/12','2026/1/13','2026/111/14'])
pd.to_datetime(s,errors = 'coerce')
```

```python
pd.to_datetime(s,errors = 'coerce').dt.month
```

```python
pd.to_datetime(s,errors = 'coerce').dt.year
```
