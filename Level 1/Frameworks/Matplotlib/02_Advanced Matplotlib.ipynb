## Colored Scatterplots

```python
iris = pd.read_csv(r'D:\Study\Data Science\Level 1\Frameworks\Matplotlib\Datasets\iris.csv')
iris.sample(5)
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa' : 0 , 'Iris-versicolor' : 1 , 'Iris-virginica' : 2})
```

```python
plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c = iris['Species'])
plt.title("Sepal And Petal")
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')
plt.show()
```

```python
# cmap => some color palletes like viridius , magma 
plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c = iris['Species'] , cmap = 'magma')
plt.title("Sepal And Petal")
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')
plt.show()
```

```python
# colorbar => shows intensity
plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c = iris['Species'])
plt.title("Sepal And Petal")
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')
plt.colorbar()
plt.show()
```

```python
# alpha = > color transpereb=ncy
plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'] , c = iris['Species'] , alpha = 0.1)
plt.title("Sepal And Petal")
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')
plt.show()
```

```python
batter.head()
```

```python
plt.figure(figsize = (15 , 7))
plt.scatter(batter['avg'] , batter['strike_rate'])
plt.title('Avg Vs Strike rate')
plt.xlabel('Avg')
plt.ylabel('Strike Rate')
plt.show()
```

```python
# basic things of fig and ax
fig , ax = plt.subplots(figsize = (8,6))
ax.scatter(batter['avg'] , batter['strike_rate'])
ax.set_title('Avg Vs Scatter')
ax.set_xlabel('Avg')
ax.set_ylabel('strike_rate')
fig.show()
```

```python
# right way to plot differnt plots in same figure
fig , ax = plt.subplots(nrows=2 , ncols=1 , sharex=True)
ax[0].scatter(batter['avg'] , batter['strike_rate'] , color = 'red')
ax[1].scatter(batter['avg'] , batter['runs'])

ax[0].set_title('Avg Vs Strike Rate')
ax[1].set_title('Avg Vs Runs')

ax[0].set_ylabel('Strike rate')
ax[1].set_ylabel('runs')
ax[1].set_xlabel('Avg')
```

```python
# way to using pyplot and creating multiple plots in same figure
fig , ax = plt.subplots(nrows=2 , ncols=2)

ax[0,0].scatter(batter['avg'] , batter['strike_rate'] , color = 'red')
ax[0,1].scatter(batter['avg'] , batter['runs'])
ax[1,0].hist(batter['avg'])
ax[1,1].hist(batter['strike_rate'] )
fig.show()
```

```python
# another way but prefer above way
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batter['avg'] , batter['strike_rate'])

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batter['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batter['avg'])

ax3 = fig.add_subplot(2,2,4)
ax3.hist(batter['strike_rate'])
```

### 3D surface plot()

```python
x = np.linspace(-10 , 10 ,20)
y = np.linspace(-10,10,20)
```

```python
xx , yy = np.meshgrid(x , y)
```

```python
z = xx**2 + yy**2
```

```python
fig = plt.figure()
ax = plt.subplot(projection = '3d')
p = ax.plot_surface(xx , yy , z , cmap= 'viridis')
fig.colorbar(p)
```

```python
z= np.sin(xx) + np.sum(yy)
```

```python
fig = plt.figure()
ax = plt.subplot(projection = '3d')
p = ax.plot_surface(xx , yy , z , cmap= 'viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```
