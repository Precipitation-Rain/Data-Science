### **`Problem-1:`** Draw a line plot of which, the x-axis is the "Year" and the y-axis is sum of "PM2.5" of two countries Iran and China.

```python
df = pd.read_csv(r'D:\Study\Data Science\Level 1\Frameworks\Matplotlib\Datasets\List of most-po01_Task_lluted cities by particulate matter concentration - List of most-polluted cities by particulate matter concentration.csv')
df
```

```python
new_df = df[(df['Country'] == 'China') | (df['Country'] == 'Iran')]
```

```python
temp = new_df.groupby(['Country' , 'Year'])['PM2.5'].sum().reset_index()
```

```python
temp
```

```python
china = temp[temp['Country'] == 'China']
```

```python
iran = temp[temp['Country'] == 'Iran']
```

```python
plt.plot(china['Year'] , china['PM2.5'] , label = 'China')
plt.plot(iran['Year'] , iran['PM2.5'] , label = 'Iran')
plt.title('Iran and China Pollution Comparisison')
plt.ylabel('PM2.5')
plt.xlabel('Year')
plt.legend()
plt.show()
# plt.ylim(30,2000)
```
