## plotly express

```python
import plotly.express as px
import pandas as pd
import numpy as np
```

```python
df = px.data.iris()
```

```python
df
```

```python
# scatter chart
fig = px.scatter(
    data_frame=df,
    x = 'sepal_length',
    y = 'sepal_width',
    color = 'species'
)
fig.show()
```

```python
# Use cases:
# Used for relationships between variables.
# Correlation
# Clustering
# Outliers
```

```python
df2 = px.data.stocks()
```

```python
df2
```

```python
# line chart => Mostly for time-series data.
# Key idea:Multiple columns → multiple lines automatically
fig = px.line(
    df2 ,
    x = 'date' ,
    y=['GOOG' , 'AAPL' , 'AMZN']
)
fig.show()
```

```python
# bar chart => For categorical comparisons
# Common options:

# barmode="group"

# barmode="stack"

fig = px.bar(
    df,
    x = 'species',
    y = 'petal_length',
    color = 'species'

)
fig.show()
```

```python
# histogram => For distributions.
fig = px.histogram(
    df,
    x = 'sepal_length',
    color = 'species',
    nbins = 20
)
fig.show()
```

```python
# box plot => For spread & outliers.
fig = px.box(
    df , 
    x = 'species',
    y = 'petal_width',
    color = 'species'
)
fig.show()
```

```python
# universal plotly parameters
fig = px.scatter(
    df , 
    x = 'sepal_length' , 
    y = 'sepal_width',
    color ='species',
    size = 'petal_width',
    hover_data=['petal_length'],
    title ='Flower Analysis'
)
fig.show()
```

```python
# labels lets you rename things shown on the chart without changing your DataFrame.
# It controls:
# Axis titles
# Legend titles
# Hover labels
# Color bar titles
```

```python
# labels
fig = px.scatter(
    df,
    x="sepal_width",
    y="sepal_length",
    color="species",
    size="petal_length",
    labels={
        "sepal_width": "Sepal Width (cm)",
        "sepal_length": "Sepal Length (cm)",
        "species": "Species",
        "petal_length": "Petal Length (cm)"
    }
)
fig.show()
```

```python
# color bar => A color bar is shown only when color is a numeric column, not categorical.
import plotly.express as px

df = px.data.iris()

fig = px.scatter(
    df,
    x="sepal_width",
    y="sepal_length",
    color="petal_length",  # numeric column
    color_continuous_scale='viridis'
)
fig.show()
```

## Styling and Layout

```python
df
```

```python
# title and its position
fig = px.scatter(
    df,
    x = 'sepal_length',
    y = 'sepal_width'
)
fig.update_layout(
    title = 'SepalLength Vs Sepal Width',
    title_x = 0.5,
    title_y = 0.95,


)
fig.show()
```

```python
# axis names
fig = px.scatter(
    df,
    x = 'sepal_length',
    y = 'sepal_width'
)
fig.update_layout(
    title = 'SepalLength Vs Sepal Width',
    title_x = 0.5,
    title_y = 0.95,

    xaxis_title='Sepal Length (cm)',
    yaxis_title = 'Sepal Width (cm)'

)
# you can also control ranges of x and y axis
fig.update_xaxes(range = [5,8])
fig.update_yaxes(range = [2,5])
fig.show()
```

```python
# font and text size
fig = px.scatter(
    df,
    x = 'sepal_length',
    y = 'sepal_width'
)
fig.update_layout(
    title = 'SepalLength Vs Sepal Width',
    title_x = 0.5,
    title_y = 0.95,

    xaxis_title='Sepal Length (cm)',
    yaxis_title = 'Sepal Width (cm)',

    font = dict(family = 'Ariel',size = 18),
    title_font_size = 20

)
fig.show()
```

```python
# backdround color and grid
fig = px.scatter(
    df,
    x = 'sepal_length',
    y = 'sepal_width'
)
fig.update_layout(
    title = 'SepalLength Vs Sepal Width',
    title_x = 0.5,
    title_y = 0.95,

    xaxis_title='Sepal Length (cm)',
    yaxis_title = 'Sepal Width (cm)',

    font = dict(family = 'Ariel',size = 18),
    
    plot_bgcolor = 'cyan',
    paper_bgcolor = 'green'

)
fig.show()
```

```python
# gridlines

fig = px.scatter(
    df,
    x = 'sepal_length',
    y = 'sepal_width'
)
fig.update_layout(
    title = 'SepalLength Vs Sepal Width',
    title_x = 0.5,
    title_y = 0.95,


    xaxis_title='Sepal Length (cm)',
    yaxis_title = 'Sepal Width (cm)',

    font = dict(family = 'Ariel',size = 18),
    
      
    plot_bgcolor = 'cyan',
    paper_bgcolor = 'green',
)

fig.update_xaxes(showgrid = True , gridcolor = 'black')
fig.update_yaxes(showgrid = True , gridcolor = 'Yellow')
fig.show()
```

```python
# legend control


fig = px.scatter(
    df,
    x = 'sepal_length',
    y = 'sepal_width'
)
fig.update_layout(
    title = 'SepalLength Vs Sepal Width',
    title_x = 0.5,
    title_y = 0.95,


    xaxis_title='Sepal Length (cm)',
    yaxis_title = 'Sepal Width (cm)',

    font = dict(family = 'Ariel',size = 18),
    
      
    plot_bgcolor = 'cyan',
    paper_bgcolor = 'green',

    legend_title_text = 'Species',
    legend = dict(
        orientation = 'h',
        yanchor = 'bottom'
    )
)

fig.update_xaxes(showgrid = True , gridcolor = 'black')
fig.update_yaxes(showgrid = True , gridcolor = 'Yellow')
fig.show()
```

## Graph Objects:

```python
import plotly.graph_objects as go
```

```python
# single trace
fig = go.Figure()

fig.add_trace(
    go.Scatter(
    x = [2,3,4,5],
    y = [5,6,7,8],
    mode = 'lines+markers',
    name = 'Line 1'
    )
)

fig.show()
```

```python
#multiple traces
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x = [1,2,3],
        y = [4,5,6],
        name = 'Line 1'
    )
)

fig.add_trace(
    go.Scatter(
        x = [7,8,9,12],
        y = [4,5,6,7],
        name = "Name 2"
    )
)

fig.show()
```

```python
# styling
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x = [1,2,3],
        y = [4,5,6],
        # name = 'Line 1',
        marker=dict(size = 14 , color = 'green' , symbol = 'circle'),
        line = dict(width = 3 , dash = 'solid'),
        name = "Styled Trace"
    )
)

fig.add_trace(
    go.Scatter(
        x = [7,8,9,12],
        y = [4,5,6,7],
        name = "Name 2"
    )
)


fig.show()
```
